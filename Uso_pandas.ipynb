#Distribuição amostral e intervalo de confiança

```python
dados.loc[dados['SALARIO'].isna(),'SALARIO']
```

```python
salarios = dados['SALARIO']
dados['SALARIO'].value_counts()
salarios.isna().sum()
len(salarios)
dados.loc[dados['SALARIO'].isna()]
dados[dados['SALARIO'].isna()]['FAIXA IDADE'].value_counts()
```

```python
media_amostral = np.mean(salarios)
desvio_amostral = np.std(salarios)

print(f"Para a presente amaostra a média é {media_amostral} e o desvio padrão é {desvio_amostral}")
```

```python
nivel_confianca = 0.95 #Arbitrário nesse caso
tamanho_amostra = len(salarios)
print(f"O tamanho da amostra selecionada é de {tamanho_amostra} pessoas")
```

```python
erro_padrao = stats.sem(salarios)
print("O erro padrão da amostra é :", erro_padrao)
```

```python
print(salarios)
print(type(salarios))
print(salarios.isna().sum())
print(salarios.dtype)
print(len(salarios))
```

```python
salarios.isna().sum()
```

```python
dados[dados['SALARIO'].isna()]['SALARIO']
```

```python
#Aparentemente existem valores (total de 577) NaN dentro da lista salarios, o que bagunça o resultado da função stats.sem
#Para ignorar esses valores NaN, utiliza-se a função do módulo stats da seguinte maneira:
#A opção por ignorar pareceu ser a mais adequada, uma vez que os dados de faixa de salário também não foram preenchidos, assim como, uma série de outras informações não foram atribuídas.
erro_padrao = stats.sem(salarios.dropna())
print("O erro padrão da amostra é :", erro_padrao)
```

```python
#O intervalo de confiança, para o qual a média obtida é válida, considerando um nível de confiança de 95%:
intervalo_confianca = stats.t.interval(nivel_confianca, tamanho_amostra - 1, loc = media_amostral, scale = erro_padrao)
print("O intervalo de confiança para essa amostra é: ", intervalo_confianca)
```

```python
def cramer_coeficiente(coluna1,coluna2):
  tabela_cruzada = np.array(pd.crosstab(coluna1, coluna2))
  chi2 = chi2_contingency(tabela_cruzada)[0] #O índice "0" indica que o único valor desejado é aquele que é oferecido primeiro pela função, que corresponde exatamente ao chi².
  soma = np.sum(tabela_cruzada)
  mini = min(tabela_cruzada.shape) - 1
  cramer = np.sqrt(chi2/(soma*mini))
  return cramer
```

```python
# A query abaixo está executando:
# busca na tabela Municipios_Brasileiros os estados correspondentes à renda informada na tabela Municipio_Status. Para isso, utiliza-se da correspondência entre as duas que é a coluna municipio_ID. Essa correspondência é garantida pelo comando INNER JOIN
# além disso, deseja-se que os estados citados sejam apenas aqueles presentes no banco de dados estraído de analise_dados.csv, que gerou a lista_UF. Para isso, utiliza-se do comando WHERE
# a função ".format" é utilizada para fazer com que sejam atribuídos, a cada linha da query, um ponto de interrogação ' ?'. Esse ponto será substituído pelos estados que aparecem em lista_UF
# Por fim, GROUP BY fará o agrupamento dos dados pelos estados listados em lista_UF

query = """SELECT Municipios_Brasileiros.Estado, AVG(Municipio_Status.renda) FROM Municipios_Brasileiros
INNER JOIN Municipio_Status ON Municipios_Brasileiros.municipio_ID = Municipio_Status.municipio_ID
WHERE Municipios_Brasileiros.Estado IN ({})
GROUP BY Municipios_Brasileiros.Estado""".format(' ,'.join(['?' for _ in lista_UF]))
```

```python
#Carrega o arquivo sql:
estados_renda = pd.read_sql(query, con=conexao, params=lista_UF)
display(estados_renda)
```
